## Vibe Coding Guidelines

### Step 1: Plan Before You Code!
**Use Cursor's todo feature to create your plan.** Here's how:

1. **Ask your AI:** "Help me create a comprehensive plan for cleaning this salary dataset and answering the business questions"
2. **Break it down:** Convert the AI's suggestions into specific, actionable todos
3. **Use Cursor's todo system** to track your progress

**Example todos you might create:**
```
- [ ] Load and explore the dataset structure
- [ ] Identify missing values and inconsistencies 
- [ ] Clean salary data (remove non-numeric characters, handle ranges)
- [ ] Standardize job titles and industries
- [ ] Clean location data (standardize state/country names)
- [ ] Handle experience data inconsistencies
- [ ] Create analysis-ready dataset
- [ ] Answer each business question with clean data
```

### Step 2: Prompt Like a Pro

**Good prompts get good results!** Use these prompting strategies:

#### Effective Prompts:
- "Show me how to identify and handle missing values in the salary column"
- "Write code to standardize job titles that contain 'software engineer', 'developer', or 'programmer'"
- "Help me extract numeric salary values from text that might contain ranges like '$50,000 - $60,000'"
- "Create a function to clean location data and extract just the US state names"

#### Avoid These Prompts:
- "Clean my data" (too vague)
- "Fix everything" (not specific enough)
- "Make it work" (no clear direction)

### Step 3: Work in Intervals

**Don't try to do everything at once!** Follow this rhythm:

1. **Code one small piece** (5-10 lines max)
2. **Run it and check the output**
3. **Sanity check:** Does this make sense?
4. **Move to the next piece**

**Sanity Check Examples:**
- After cleaning salaries: "Do the salary values look reasonable? Are they in the right range?"
- After standardizing locations: "Are the state names consistent? Any weird values left?"
- After each analysis: "Does this number make business sense?"

### Step 4: Document Everything

**Your future self (and your manager) will thank you!**

- Add markdown cells explaining each major step
- Comment your code to explain the "why" not just the "what"
- Include your todo plan in the notebook
- Note any assumptions or decisions you made
